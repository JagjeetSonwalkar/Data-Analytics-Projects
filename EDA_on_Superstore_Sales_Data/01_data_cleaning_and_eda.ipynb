# Checking Missing & Duplicate Values

```python
# Missing Value of Orders DF
Total_Missing_Value = orders_df.isna().sum().sum()
Missing_value_Percent = np.round((orders_df.isna().sum().sum() / orders_df.shape[0] ) * 100, 2)
Total_Duplicate_value = orders_df.duplicated(subset=("Order ID","Customer ID","Product ID","Product Name")).sum()

print(f"Total Missing Value in Order DF: {Total_Missing_Value}")
print(f"Missing Value Percentage % in Order DF: {Missing_value_Percent}%")
print(f"Total Duplicate Values in Order DF: {Total_Duplicate_value}")
```

```python
# Missing Value of People DF
Total_Missing_Value = people_df.isna().sum().sum()
Missing_value_Percent = np.round((people_df.isna().sum().sum() / people_df.shape[0] ) * 100, 2)
Total_Duplicate_value = people_df.duplicated(subset = ("Person","Region")).sum()

print(f"Total Missing Value in people DF: {Total_Missing_Value}")
print(f"Missing Value Percentage % in people DF: {Missing_value_Percent}%")
print(f"Total Duplicate Values in people DF: {Total_Duplicate_value}")
```

```python
# Missing Value of Return DF
Total_Missing_Value = return_df.isna().sum().sum()
Missing_value_Percent = np.round((return_df.isna().sum().sum() / return_df.shape[0] ) * 100, 2)
Total_Duplicate_value = return_df.duplicated(subset=("Column1","Column2")).sum()

print(f"Total Missing Value in return DF: {Total_Missing_Value}")
print(f"Missing Value Percentage % in return DF: {Missing_value_Percent}%")
print(f"Total Duplicate Values in return DF: {Total_Duplicate_value}")
```

# Visualization

```python
# Sales by Category
category_sales = (
    orders_df.groupby("Category")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

category_sales.plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")
plt.xticks(rotation=0)
plt.show()
```

```python
# Sales by Sub-Category
subcategory_sales = (
    orders_df.groupby("Sub-Category")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

subcategory_sales.plot(
    kind="bar",
    figsize=(12, 6)
)

plt.title("Total Sales by Sub-Category")
plt.xlabel("Sub-Category")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.show()
```

```python
# Sales by Segment
segment_sales = (
    orders_df.groupby("Segment")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

segment_sales.plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Total Sales by Segment")
plt.xlabel("Segment")
plt.ylabel("Sales")
plt.xticks(rotation=0)
plt.show()
```

```python
# Sales by Region
region_sales = (
    orders_df.groupby("Region")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

region_sales.plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Total Sales by Region")
plt.xlabel("Region")
plt.ylabel("Sales")
plt.xticks(rotation=0)
plt.show()
```

```python
# Sales by State
state_sales = (
    orders_df.groupby("State")["Sales"]
    .sum()
    .sort_values(ascending=True)
)

state_sales.plot(
    kind="barh",
    figsize=(10, 12)
)

plt.title("Total Sales by State")
plt.xlabel("Sales")
plt.ylabel("State")
plt.show()
```

```python
# Sales vs Profit
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=orders_df,
    x="Sales",
    y="Profit"
)

plt.title("Sales vs Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.show()
```

```python
 plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=orders_df,
    x="Sales",
    y="Profit",
    hue="Category"
)

plt.title("Sales vs Profit by Category")
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.show()
```

```python
# Discount vs Profit
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=orders_df,
    x="Discount",
    y="Profit"
)

plt.title("Discount vs Profit")
plt.xlabel("Discount")
plt.ylabel("Profit")
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=orders_df,
    x="Discount",
    y="Profit",
    hue="Category"
)

plt.title("Discount vs Profit by Category")
plt.xlabel("Discount")
plt.ylabel("Profit")
plt.show()
```

```python
# Sales Distribution
plt.figure(figsize=(10, 6))

sns.histplot(
    data=orders_df,
    x="Sales",
    bins=30
)

plt.title("Distribution of Sales")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```

```python
# Profit Distribution
plt.figure(figsize=(10, 6))

sns.histplot(
    data=orders_df,
    x="Profit",
    bins=30
)

plt.title("Distribution of Profit")
plt.xlabel("Profit")
plt.ylabel("Frequency")
plt.show()
```

```python
# Quantity Distribution
plt.figure(figsize=(10, 6))

sns.histplot(
    data=orders_df,
    x="Quantity",
    bins=10
)

plt.title("Distribution of Quantity")
plt.xlabel("Quantity")
plt.ylabel("Frequency")
plt.show()
```

```python
# Discount Distribution
plt.figure(figsize=(10, 6))

sns.histplot(
    data=orders_df,
    x="Discount",
    bins=20
)

plt.title("Distribution of Discount")
plt.xlabel("Discount")
plt.ylabel("Frequency")
plt.show()
```

```python
# Correlation Heatmap
numeric_cols = [
    "Sales",
    "Profit",
    "Quantity",
    "Discount"
]

corr = orders_df[numeric_cols].corr()

plt.figure(figsize=(8, 6))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
# Outlier Detection: Use boxplots to identify outliers in Sales, Profit, and Discount

fig, axes = plt.subplots(3, 1, figsize=(10, 10))

sns.boxplot(data=orders_df, x="Sales", ax=axes[0])
axes[0].set_title("Sales")
axes[0].set_xlabel("")

sns.boxplot(data=orders_df, x="Profit", ax=axes[1])
axes[1].set_title("Profit")
axes[1].set_xlabel("")

sns.boxplot(data=orders_df, x="Discount", ax=axes[2])
axes[2].set_title("Discount")
axes[2].set_xlabel("")

plt.tight_layout()
plt.show()
```
