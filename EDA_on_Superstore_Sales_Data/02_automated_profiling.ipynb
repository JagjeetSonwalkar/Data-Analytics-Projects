```python
import pandas as pd
from ydata_profiling import ProfileReport
```

```python
orders_df = pd.read_excel("Data/sample_superstore.xlsx", sheet_name="Orders")
people_df = pd.read_excel("Data/sample_superstore.xlsx", sheet_name="People")
return_df = pd.read_excel("Data/sample_superstore.xlsx", sheet_name="Returns")
```

```python
orders_df_prof = ProfileReport(orders_df)
orders_df_prof.to_file(output_file = 'reports/ydata_profiling_output/orders_df_Data_profiling.html')
```

```python
people_df_prof = ProfileReport(people_df)
people_df_prof.to_file(output_file = 'reports/ydata_profiling_output/people_df_Data_profiling.html')
```

```python
return_df_prof = ProfileReport(return_df)
return_df_prof.to_file(output_file = 'reports/ydata_profiling_output/return_df_Data_profiling.html')
```
